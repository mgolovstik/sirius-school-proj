```python
# Ячейка 3: Имитация API запросов
async def mock_api_call(lat, lon, api_type="maps"):
    """Заглушка для API запросов"""
    await asyncio.sleep(0.1)  # Имитация задержки
    
    if api_type == "maps":
        return {
            "poi_count": np.random.randint(10, 100),
            "metro_distance": np.random.randint(100, 2000),
            "district": np.random.choice(["Центр", "СВАО", "ЮЗАО"])
        }
    else:
        return {
            "address": f"Москва, {lat}, {lon}",
            "rubrics": ["business", "office"]
        }

# Тест асинхронных запросов
async def test_async():
    tasks = [
        mock_api_call(55.75, 37.61, "maps"),
        mock_api_call(55.75, 37.61, "yandex")
    ]
    results = await asyncio.gather(*tasks)
    return results

# Запуск теста
results = await test_async()
print(results)
```
