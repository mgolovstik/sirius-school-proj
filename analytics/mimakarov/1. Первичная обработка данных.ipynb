```python
data["Название"] = data["Название"].fillna("").astype(str)
data["Описание"] = data["Описание"].fillna("").astype(str)
data["Описание"] = "НАЗВАНИЕ: " + data["Название"] + "\n\n" + "ОПИСАНИЕ: " + data["Описание"]
```

```python
data = data.drop("Название", axis=1)
```

```python
data["Доп.параметры"].iloc[10]
```

```python
def parse_parameters(v):
    return dict(
        pair.split("=")
        for pair in v.split("|")
    )

parse_parameters(data["Доп.параметры"].iloc[10])
```

```python
_ = data["Доп.параметры"].apply(parse_parameters)
params_parsed = pd.json_normalize(_)
params_parsed.index = data.index
params_parsed
```

```python
params_parsed.isna().sum() / params_parsed.shape[0] * 100
```

```python
params_parsed.nunique()
```

```python
params_parsed["Ед. измерения площади"].unique()
```

```python
params_parsed = params_parsed.drop(["Класс здания", "Тип объявления"], axis=1)
```

```python
data = pd.concat(
    [
        data.drop("Доп.параметры", axis=1),
        params_parsed
    ],
    axis=1
)
data
```
