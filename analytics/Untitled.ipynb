```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("../data/data_part5.csv")
data.head()
```

```python
data.columns
```

```python
data = data.drop(['Телефон', 'Оператор', 'Контактное лицо (автор объявления)', 'Регион мобильного телефона', 'Номер подменён', 'Тип объявления', 'Категория1',
       'Категория2', 'Персона для контактов'], axis=1)
```

```python
data = data.set_index(['ID на сайте', 'Источник'])
```

```python
data.head()
```

```python
data.loc[(314143748, "cian.ru")]
```

```python
data["Описание"] = "НАЗВАНИЕ: " + data["Название"] + '\n\n' + "ОПИСАНИЕ: " + data["Описание"]
data = data.drop("Название", axis = 1)
print(data.loc[(314143748, "cian.ru"), "Описание"])
```

```python
data["Доп.параметры"].iloc[0]
```

```python
def parse_param(v):
    return dict(pair.split("=") for pair in v.split("|"))

parse_param(data["Доп.параметры"].iloc[0])
```

```python
_ = data["Доп.параметры"].apply(parse_param)
params_parsed = pd.json_normalize(_)
params_parsed.index = data.index
```

```python
params_parsed.isna().sum() / params_parsed.shape[0] * 100
```

```python
params_parsed.nunique()
```

```python
params_parsed["Ед. измерения площади"].unique()
```

```python
params_parsed.drop(["Класс здания", "Тип объявления"], axis = 1)
```

```python
data = pd.concat([data.drop("Доп.параметры", axis = 1), params_parsed], axis = 1)
```

```python
data["Дата"] = pd.to_datetime(data["Дата"])
```

```python
data[["Расстояние до метро, км", "Этаж", "Общая площадь", "Этажность здания"]]
```

```python
data["Расстояние до метро, км"] = data["Расстояние до метро, км"].replace({"неизвестно": np.nan})
data[["Расстояние до метро, км", 
      "Этаж", 
      "Общая площадь", 
      "Этажность здания"]] = data[[
    "Расстояние до метро, км", 
    "Этаж", 
    "Общая площадь", 
    "Этажность здания"]].astype(float)
```

```python
data.loc[data["Ед. измерения площади"] == "сот.", "Общая площадь"] *= 100
data.loc[data["Ед. измерения площади"] == "га.", "Общая площадь"] *= 10000
```

```python
data["Цена / м2"] = data["Цена"] / data["Общая площадь"]
```

```python
data = data.drop(["Регион", "Класс здания", "Ед. измерения площади", "Тип объявления", "Категория"], axis = 1)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = data[data["Вид объекта"] != "Коммерческая земля"].copy()
data = data[data["Вид объекта"] != "Складское помещение"].copy()
data = data[data["Вид объекта"] != "Производственное помещение"].copy()
data = data[data["Вид объекта"] != "Гостиница"].copy()
data = data[data["Общая площадь"] <= 2000].copy()
data = data[data["Общая площадь"] >= 20].copy()
```

```python
data.shape
```

```python
# data = data[data["Цена"] <= 3e6].copy()
# data = data[data["Цена"] >= 20000].copy()
```

```python
data = data.dropna(subset = ["Вид объекта"])
```

```python
data.shape
```

```python
mask = data["Цена / м2"].quantile(0.95)
data.loc[data["Цена / м2"] > mask, "Цена"] //= 12
```

```python
data = data[(data["Цена / м2"] <= 10000) & (data["Цена / м2"] >= 200)]
```

```python
data.shape
```

```python
data["Вид объекта"] = data["Вид объекта"].replace({"Помещение свободного назначения": "Торговое/Свободного назначения",
                             "Торговое помещение": "Торговое/Свободного назначения",
                             "Здание": "Офис/Здание",
                             "Офисное помещение": "Офис/Здание"})
```

```python
data.to_csv("../data/data_part5_t.csv")
```
