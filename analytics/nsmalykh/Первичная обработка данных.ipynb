```python
VAL = 500000

# Фильтруем цены до VAL
filtered_prices = data[data["Цена"] <= VAL]["Цена"]

plt.figure(figsize=(12, 6))
plt.hist(filtered_prices, bins=50, edgecolor='black', alpha=0.7)
plt.xlabel(f'Цена (до {VAL:,.0f})')
plt.ylabel('Частота')
plt.title(f'Распределение цен до {VAL:,.0f} (всего {len(filtered_prices)} объектов)')
plt.grid(True, alpha=0.3)

# Добавляем вертикальную линию для средней цены
if len(filtered_prices) > 0:
    mean_price = filtered_prices.mean()
    plt.axvline(mean_price, color='red', linestyle='--', linewidth=2, 
                label=f'Средняя цена: {mean_price:,.0f}')
    plt.legend()

plt.show()

# Выводим статистику
print(f"Всего объектов с ценой до {VAL:,.0f}: {len(filtered_prices)}")
print(f"Из общего количества: {len(data)}")
print(f"Процент: {len(filtered_prices)/len(data)*100:.1f}%")
if len(filtered_prices) > 0:
    print(f"\nСтатистика по ценам до {VAL:,.0f}:")
    print(f"Средняя цена: {filtered_prices.mean():,.0f}")
    print(f"Медианная цена: {filtered_prices.median():,.0f}")
    print(f"Минимальная цена: {filtered_prices.min():,.0f}")
    print(f"Максимальная цена: {filtered_prices.max():,.0f}")
```

```python
# Если в графике: x=lat, y=lng
# Тогда lat - это координата X, lng - координата Y

x = 55.658850  # lat (широта) - ось X
y = 37.197321  # lng (долгота) - ось Y
EPS = 0.04

mask1 = abs(data["lat"] - x) < EPS   # Сравниваем lat с X
mask2 = abs(data["lng"] - y) < EPS   # Сравниваем lng с Y

filtered_data = data[mask1 & mask2]

# Проверяем, есть ли столбец с ценами
if 'Цена' not in filtered_data.columns:
    print("Столбец 'Цена' не найден в данных.")
    print("Доступные столбцы:", list(filtered_data.columns))
else:
    # Вычисляем среднюю цену
    avg_price = filtered_data["Цена"].mean()
    
    # Более подробная статистика
    price_stats = filtered_data["Цена"].describe()
    
    print(f"Найдено {len(filtered_data)} точек рядом с ({x:.6f}, {y:.6f}):")
    print("=" * 80)
    
    # Выводим информацию о каждой точке
    for idx, (_, row) in enumerate(filtered_data.iterrows(), 1):
        print(f"{idx}. Координаты: lat={row['lat']:.6f}, lng={row['lng']:.6f}")
        print(f"   URL: {row['URL']}")
        print(f"   Цена: {row['Цена']:,.0f}" if pd.notna(row['Цена']) else "   Цена: (нет данных)")
        print("-" * 80)
    
    print("\n" + "=" * 80)
    print("СТАТИСТИКА ПО ЦЕНАМ:")
    print(f"Количество точек с указанной ценой: {filtered_data['Цена'].count()} из {len(filtered_data)}")
    print(f"Средняя цена: {avg_price:,.0f}")
    print(f"Медианная цена: {filtered_data['Цена'].median():,.0f}")
    print(f"Минимальная цена: {filtered_data['Цена'].min():,.0f}")
    print(f"Максимальная цена: {filtered_data['Цена'].max():,.0f}")
    print(f"Стандартное отклонение: {filtered_data['Цена'].std():,.0f}")
```
